```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

from sklearn.neighbors import NearestNeighbors

data1 = pd.read_csv('homework_6.1.csv')
```

```python
data1.head()
```

### Average Treatment Effect on Treated

```python
nbrs = NearestNeighbors(n_neighbors = 1)

X_0 = data1[data1['X'] == 0]
X_1 = data1[data1['X'] == 1]

nbrs.fit(X_0[['Z']])

distances, indices = nbrs.kneighbors(X_1[['Z']])
```

```python
matched_X0 = X_0.iloc[indices.flatten()]

avg_y_1 = X_1['Y'].mean()
avg_y_0 = matched_X0['Y'].mean()

effect_treated = avg_y_1 - avg_y_0

print(f'The average treated effect is {effect_treated:.3f}')
```

### Average Treatment Effect on Untreated

```python
nbrs.fit(X_1[['Z']])

distances, indices = nbrs.kneighbors(X_0[['Z']])

matched_X1 = X_1.iloc[indices.flatten()]

avg_y_1 = matched_X1['Y'].mean()
avg_y_0 = X_0['Y'].mean()

effect_untreated = avg_y_1 - avg_y_0

print(f'The average untreated effect is {effect_untreated:.3f}')
```

### Average Treatment Effect

```python
untreated_effects = matched_X1['Y'].to_numpy() - X_0['Y'].to_numpy()
treated_effects = X_1['Y'].to_numpy() - matched_X0['Y'].to_numpy()

all_effects = np.concatenate([untreated_effects, treated_effects])

ate = all_effects.mean()

print(f'The average treatment effect is {ate:.3f}')
```

### Optimal Treatment Effect

```python
optimal_effect = untreated_effects.max()

print(f'The optimal treatment effect is {optimal_effect:.3f}')
```
