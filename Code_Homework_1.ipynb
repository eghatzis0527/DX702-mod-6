```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data_1 = pd.read_csv("homework_1.1.csv")

data_1.head()
```

```python
# Multiple Regression on all Xs
import statsmodels.api as sm

X = data_1.drop(columns = ['Y'])
X = sm.add_constant(X)
Y = data_1['Y']

model = sm.OLS(Y, X).fit()

model.summary()
```

```python
# Single Regression for each Xs
X1 = data_1[['X1']]
X1 = sm.add_constant(X1)

X2 = data_1[['X2']]
X2 = sm.add_constant(X2)

X3 = data_1[['X3']]
X3 = sm.add_constant(X3)

model1 = sm.OLS(Y, X1).fit()
t_val1 = model1.tvalues
coef_1 = model1.params

model2 = sm.OLS(Y, X2).fit()
t_val2 = model2.tvalues
coef_2 = model2.params

model3 = sm.OLS(Y, X3).fit()
t_val3 = model3.tvalues
coef_3 = model3.params

print(f"X1 T-Value: {t_val1['X1']:.2f}, X1 Coefficient: {coef_1['X1']:.2f}")
print(f'X2 T-Value: {t_val2['X2']:.2f}, X2 Coefficient: {coef_2['X2']:.2f}')
print(f'X3 T-Value: {t_val3['X3']:.2f}, X3 Coefficient: {coef_3['X3']:.2f}')
```

```python
data2 = pd.read_csv('homework_1.2.csv')

data2.head()
```

```python
# Fitting the nearest neighbors
from sklearn.neighbors import NearestNeighbors

nbrs = NearestNeighbors(n_neighbors = 1)

X_0 = data2[data2['X'] == 0]
X_1 = data2[data2['X'] == 1]

nbrs.fit(X_0[['Z']])

distances, indices = nbrs.kneighbors(X_1[['Z']])
```

```python
# Closest Distance between Matches
min(distances)
```

```python
# Furthest Distance between Matches (Question 4)
max(distances)
```

```python
# Answering Question 5
matched_X0 = X_0.iloc[indices.flatten()]

avg_y_1 = X_1['Y'].mean()
avg_y_matched0 = matched_X0['Y'].mean()

effect_a = avg_y_1 - avg_y_matched0

print(f"The effect is {effect_a:.5f}.")
```

```python
# Using radius neighbors to get matches within 0.2
nbrs_b = NearestNeighbors(radius = 0.2)

nbrs_b.fit(X_0[['Z']])

distances, indices = nbrs_b.radius_neighbors(X_1[['Z']])
```

```python
# Getting the number of duplicates

all_indices = np.concatenate(indices)

duplicates = pd.Series(all_indices).value_counts()

print(f"Number of duplicates: {sum(duplicates) - 52}")
```

```python
# Calculating the Effect of Approach B

approach_b_means = []

for i in indices:
    if len(i) > 0: 
        approach_b_means.append(X_0.iloc[i]['Y'].mean())

approach_b_means[0:5]
```

```python
overall_mean = np.mean(approach_b_means)

effect_b = avg_y_1 - overall_mean

print(f"The effect for Approach B is {effect_b:.4f}.")
```
