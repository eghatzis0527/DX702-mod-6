```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data_1 = pd.read_csv("homework_1.1.csv")

data_1.head()
```

```python
from sklearn.linear_model import LinearRegression

X = data_1.drop(columns = ['Y'])
Y = data_1['Y']

model = LinearRegression()

model.fit(X, Y)

preds = model.predict(X)

model.coef_
```

```python
X1 = data_1[['X1']]
X2 = data_1[['X2']]
X3 = data_1[['X3']]

model.fit(X1, Y)
preds_1 = model.predict(X1)
coef_1 = model.coef_

model.fit(X2, Y)
preds_2 = model.predict(X2)
coef_2 = model.coef_

model.fit(X3, Y)
preds_3 = model.predict(X3)
coef_3 = model.coef_

print(coef_1)
print(coef_2)
print(coef_3)
```
