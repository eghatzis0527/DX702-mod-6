```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

data1 = pd.read_csv('homework_4.1.csv')
```

```python
data1.head()
```

```python
y_mean1 = data1.loc[data1['Z'] == 1, 'Y'].mean()
y_mean0 = data1.loc[data1['Z'] == 0, 'Y'].mean()

x_mean1 = data1.loc[data1['Z'] == 1, 'X'].mean()
x_mean0 = data1.loc[data1['Z'] == 0, 'X'].mean()

effect1 = (y_mean1 - y_mean0)/(x_mean1 - x_mean0)

print(f"The first calculated effect is: {effect1 :.1f}.")
```

```python
data1['w_range'] = pd.cut(data1['W'], bins = 10)

effects = []

for w_range, group in data1.groupby('w_range', observed = True):

    y_mean1 = group.loc[group['Z'] == 1, 'Y'].mean()
    y_mean0 = group.loc[group['Z'] == 0, 'Y'].mean()

    x_mean1 = group.loc[group['Z'] == 1, 'X'].mean()
    x_mean0 = group.loc[group['Z'] == 0, 'X'].mean()

    y_diff = y_mean1 - y_mean0
    x_diff = x_mean1 - x_mean0

    if x_diff != 0 and pd.notna(x_diff) and pd.notna(y_diff): 
        effect = y_diff/x_diff

        effects.append(effect)

overall_effect = np.mean(effects)

print(f"The overall mean for the second task is {overall_effect :.2f}.")
```

```python
data2 = pd.read_csv('homework_4.2.a.csv')
data3 = pd.read_csv('homework_4.2.b.csv')

display(data2.head())
display(data3.head())
```

```python
data2['treated'] = (data2['X'] >= 80).astype(int)
data2['X_centered'] = data2['X'] - 80
data2['interaction'] = data2['treated'] * data2['X_centered']

data3['treated'] = (data3['X2'] >= 80).astype(int)
data3['X_centered'] = data3['X2'] - 80
data3['interaction'] = data3['treated'] * data3['X_centered']

X1 = data2[['treated', 'X_centered', 'interaction']]
X1 = sm.add_constant(X1)
y1 = data2['Y']

X2 = data3[['treated', 'X_centered', 'interaction']]
X2 = sm.add_constant(X2)
y2 = data3['Y2']

model1 = sm.OLS(y1, X1).fit()
model2 = sm.OLS(y2, X2).fit()

display(model1.summary())
display(model2.summary())
```
