```python
import pandas as pd 
import numpy as np 
import statsmodels.api as sm 
```

```python
data1 = pd.read_csv('homework_12.1.csv')
data2 = pd.read_csv('homework_12.2.csv')

display(data1.head())
display(data2.head())
```

```python
data1['Post'] = (data1['Time'] > 0).astype(int)

X = pd.DataFrame({'Group' : data1['Group'], 'Time' : data1['Post'], 'Combined' : (data1['Group'] * data1['Post'])})
X = sm.add_constant(X)

model = sm.OLS(data1['Y'], X).fit()

model.summary()
```

```python
data = data2[data2['Time'] < 0]

data['Interaction'] = data['Group'] * data['Time']

X = pd.DataFrame({'Group' : data['Group'], 'Time' : data['Time'], 'Interaction' : data['Interaction']})
X = sm.add_constant(X)

model = sm.OLS(data['Y'], X).fit()

model.summary()
```

```python
n = 10000

error = np.zeros(n)
error[0] = np.random.normal(0, 1)

for t in range(1, n):
    error[t] = 0.8 * error[t-1] + np.sqrt(1 - 0.8**2) * np.random.normal(0, 1)

X = error

error2 = np.zeros(n)
error2[0] = np.random.normal(0, 1)

for t in range(1, n):
    error2[t] = 0.8 * error2[t-1] + np.sqrt(1 - 0.8**2) * np.random.normal(0, 1)

Y = 2 * X + error2

print(f'Mean: {np.mean(error)}')
print(f'Standard Deviation: {np.std(error)}')
print(f'Correlation: {np.corrcoef(error[:-1], error[1:])[0,1]}')
```

```python
X_OLS = sm.add_constant(X)
model = sm.OLS(Y, X_OLS).fit()

print(model.params)
print(model.bse)
```

```python
estimates = []

for i in range(1000):

    n = 10000

    error = np.zeros(n)
    error[0] = np.random.normal(0, 1)

    for t in range(1, n):
        error[t] = 0.8 * error[t-1] + np.sqrt(1 - 0.8**2) * np.random.normal(0, 1)

    X = error

    error2 = np.zeros(n)
    error2[0] = np.random.normal(0, 1)

    for t in range(1, n):
        error2[t] = 0.8 * error2[t-1] + np.sqrt(1 - 0.8**2) * np.random.normal(0, 1)

    Y = 2 * X + error2

    X_ols = sm.add_constant(X)

    model = sm.OLS(Y, X_ols).fit()

    estimates.append(model.params[1])

print(np.std(estimates))
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

np.random.seed(0) 

X1 = np.random.normal(0, 1, 1000) 
X2 = np.random.normal(0, 1, 1000) + X1 
X3 = np.random.normal(0, 1, 1000) + 2 * X2 

data = pd.DataFrame({'X1' : X1, 'X2' : X2, 'X3' : X3})

vif = variance_inflation_factor(data.values, 0)

print(vif)
```
