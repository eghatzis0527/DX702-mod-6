```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
```

```python
data_1 = pd.read_csv('homework_3.1.csv')

data_1.head()
```

```python
data_1['After'] = (data_1['time'] >= 50).astype(int)
data_1['Time_After'] = (data_1['time'] - 50) * data_1['After']

data_1.head()
```

```python
X = data_1[['time', 'After']]
X = sm.add_constant(X)

y = data_1['value1']

model1 = sm.OLS(y, X).fit()

print(model1.summary())
```

```python
X = data_1[['time', 'After']]
X = sm.add_constant(X)

y = data_1['value2']

model2 = sm.OLS(y, X).fit()

print(model2.summary())
```

```python
X = data_1[['time', 'After']]
X = sm.add_constant(X)

y = data_1['value3']

model3 = sm.OLS(y, X).fit()

print(model3.summary())
```

```python
X = data_1[['time', 'After', 'Time_After']]
X = sm.add_constant(X)

y = data_1['value1']

model1 = sm.OLS(y, X).fit()

print(model1.pvalues['Time_After'])
```

```python
X = data_1[['time', 'After', 'Time_After']]
X = sm.add_constant(X)

y = data_1['value2']

model2 = sm.OLS(y, X).fit()

print(model2.pvalues['Time_After'])
```

```python
X = data_1[['time', 'After', 'Time_After']]
X = sm.add_constant(X)

y = data_1['value3']

model3 = sm.OLS(y, X).fit()

print(model3.pvalues['Time_After'])
```
