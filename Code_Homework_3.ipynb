```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
```

```python
data_1 = pd.read_csv('homework_3.1.csv')

data_1.head()
```

```python
data_1['After'] = (data_1['time'] >= 50).astype(int)
data_1['Time_After'] = (data_1['time'] - 50) * data_1['After']

data_1.head()
```

```python
X = data_1[['time', 'After']]
X = sm.add_constant(X)

y = data_1['value1']

model1 = sm.OLS(y, X).fit()

print(model1.summary())
```

```python
X = data_1[['time', 'After']]
X = sm.add_constant(X)

y = data_1['value2']

model2 = sm.OLS(y, X).fit()

print(model2.summary())
```

```python
X = data_1[['time', 'After']]
X = sm.add_constant(X)

y = data_1['value3']

model3 = sm.OLS(y, X).fit()

print(model3.summary())
```

```python
X = data_1[['time', 'After', 'Time_After']]
X = sm.add_constant(X)

y = data_1['value1']

model1 = sm.OLS(y, X).fit()

print(model1.pvalues['Time_After'])
```

```python
X = data_1[['time', 'After', 'Time_After']]
X = sm.add_constant(X)

y = data_1['value2']

model2 = sm.OLS(y, X).fit()

print(model2.pvalues['Time_After'])
```

```python
X = data_1[['time', 'After', 'Time_After']]
X = sm.add_constant(X)

y = data_1['value3']

model3 = sm.OLS(y, X).fit()

print(model3.pvalues['Time_After'])
```

```python
data2 = pd.read_csv('homework_3.2.a.csv')
data3 = pd.read_csv('homework_3.2.b.csv')

display(data2.head())

display(data3.head())
```

```python
data2['interaction'] = data2['group1'] * data2['time1']

data3['interaction'] = data3['group2'] * data3['time2']

display(data2.head())
display(data3.head())
```

```python
X2 = data2[['group1', 'time1', 'interaction']]
X2 = sm.add_constant(X2)
y2 = data2['outcome1']

X3 = data3[['group2', 'time2', 'interaction']]
X3 = sm.add_constant(X3)
y3 = data3['outcome2']

model2 = sm.OLS(y2, X2).fit()
model3 = sm.OLS(y3, X3).fit()

print(f'The estimated treatment effect for dataset A is {model2.params['interaction']:.2f}')
print(f'The estimated treatment effect for dataset B is {model3.params['interaction']:.2f}')
```

```python
display(model2.summary())
display(model3.summary())
```
