```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

data1 = pd.read_csv("homework_2.1.csv")
```

```python
data1.head()
```

```python
X1 = data1['G1']

X2 = data1['G2']

X3 = data1['G3']

y = data1["time"]
y = sm.add_constant(y)

model1 = sm.OLS(X1, y).fit()
coef1 = model1.params

model2 = sm.OLS(X2, y).fit()
coef2 = model2.params

model3 = sm.OLS(X3, y).fit()
coef3 = model3.params

print(f"G1 Coefficient: {coef1.iloc[1]:.5f}")
print(f"G2 Coefficient: {coef2.iloc[1]:.5f}")
print(f"G3 Coefficient: {coef3.iloc[1]:.5f}")
```

```python
x_mean = data1['time'].mean()

y_means = {'G1' : data1['G1'].mean(), 'G2' : data1['G2'].mean(), 'G3' : data1['G3'].mean()}

y_values = []
x_values = []

for i in range(3):
    x_values.extend(data1['time'] - x_mean)

for i in y_means.keys():
    y_values.extend(data1[i] - y_means[i])

print(len(x_values))
print(len(y_values))
```

```python
data = {'time' : x_values, 'groups' : y_values}

final_data = pd.DataFrame(data)

final_data.head()
```

```python
x = final_data['time']
x = sm.add_constant(x)

y = final_data['groups']

model = sm.OLS(y, x).fit()

model.params
```

```python
data2 = pd.read_csv("homework_2.2.csv")

data2.head()
```

```python
effects = []

for i in range(1000):

    sample = data2.sample(n = len(data2), replace = True)

    treated_mean = sample[sample['X'] == 1]['Y'].mean()
    untreated_mean = sample[sample['X'] == 0]['Y'].mean()

    effects.append(treated_mean - untreated_mean)

mean_effect = np.mean(effects)
var_effect = np.var(effects, ddof = 1)

print(f'The mean effect by subtracting is {mean_effect:.3f}.')
print(f'The variance of effect by subtracting is {var_effect:.6f}.')
```

```python
from scipy.stats import skew

effect = []

for i in range(100000):

    sample = data2.sample(n = len(data2), replace = True)

    X = sample['X']
    X = sm.add_constant(X)

    y = sample['Y']

    model = sm.OLS(y, X).fit()

    effect.append(model.params['X'])

effect_skewness = skew(effect)

print(f'The skewness of the measured effect is {effect_skewness:.5f}.')
```
