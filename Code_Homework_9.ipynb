```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
```

```python
def simulate(A=1, B=1, C=10, D=1000):
  W = np.random.normal(0,1,D)
  X = W+np.random.normal(0,B,D)
  Y = A*X-W+np.random.normal(0,C,D)
  return Y, X, W

significant = 0

for i in range(1000):

  Y, X, W = simulate()

  X_data = pd.DataFrame({'X' : X, 'W' : W})
  X_data = sm.add_constant(X_data)

  model = sm.OLS(Y, X_data).fit()

  t_value = model.tvalues['X']

  if abs(t_value) > 1.96:
    significant += 1

probability = significant/1000

print(f'The probability of the t_value for X to be significant is {probability * 100:.2f}%')
```

```python
from scipy.stats import skew

coefs = []

for i in range(1000):

    Y, X, W = simulate()

    X_data = pd.DataFrame({'X' : X, 'W' : W})
    X_data = sm.add_constant(X_data)

    model = sm.OLS(Y, X_data).fit()

    coefs.append(model.params['X'])

skewness = skew(coefs)

print(f'The skew is: {skewness:.2f}')
```
