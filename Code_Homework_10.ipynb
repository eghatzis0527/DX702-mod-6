```python
import pandas as pd 
import numpy as np 
import statsmodels.api as sm 
```

```python
data1 = pd.read_csv('homework_10.1.csv')

data1.head()
```

```python
for t in range(12):

    data_t = data1[data1['time'] == t]

    X = data_t['X']
    X = sm.add_constant(X)
    y = data_t['y']

    model = sm.OLS(y, X).fit()

    print(f'For time {t} the fixed effect is {model.params['const']:.4f}')
```

```python
for c in range(10):

    data_c = data1[data1['city'] == c]

    X = data_c['X']
    X = sm.add_constant(X)
    y = data_c['y']

    model = sm.OLS(y, X).fit()

    print(f'For city {c} the fixed effect is {model.params['const']:.4f}')
```

```python
num = 10000 

X = np.clip(np.random.normal(3, 1, (num,)), 0.01, 100)
Z = np.clip(np.random.normal(3, 1, (num,)), 0.01, 100)
Y = np.log(X + Z) + np.random.normal(0, 1, (num,)) 

exp_Y = np.exp(Y)

data = pd.DataFrame({'X' : X, 'Z' : Z})
data = sm.add_constant(data)

model = sm.OLS(exp_Y, data).fit()

print(f'The coefficient for X is {model.params['X']:.4f}')
print(f'The coefficient for Z is {model.params['Z']:.4f}')
```

```python
num = 10000 

Z = np.random.normal(0, 1, (num,)) 
X = Z + np.random.normal(0, 1, (num,)) 
Y = 1.5 * X + 2.3 * Z + np.random.normal(0, X**2, (num,))

data = pd.DataFrame({'X' : X, 'Z' : Z})
data = sm.add_constant(data)

model = sm.OLS(Y, data).fit()

print(f'The estimated standard error for X is {model.bse['X']:.4f}')
```

```python
coefs = []

for i in range(100):

    num = 10000 

    Z = np.random.normal(0, 1, (num,)) 
    X = Z + np.random.normal(0, 1, (num,)) 
    Y = 1.5 * X + 2.3 * Z + np.random.normal(0, X**2, (num,))

    data = pd.DataFrame({'X' : X, 'Z' : Z})
    data = sm.add_constant(data)

    model = sm.OLS(Y, data).fit()

    coefs.append(model.params['X'])

simulated_se = np.std(coefs)

print(f'The simulated standard error is {simulated_se:.4f}')
```
